#### Task 4.1: Calculate summary statistics of numeric variables (2 points)

- Use any functions from the `tidyverse` to create **one** table that outputs the mean, minimum, and maximum of **all** numeric columns in your data, dropping the missing values if they exist.
- Show the outputted table.

</div>

```r
summary_basketball <- tidy_basketball |> # making new summary statistics data frame from tidy data frame
    #making a summary of each numeric columns in the data frame, calculating max, min, mean for each
    summarise(across(where(is.numeric),list(mean = ~mean(.x, na.rm = TRUE),min = ~min(.x, na.rm = TRUE), max = ~max(.x, na.rm = TRUE))))
    # na.rm = TRUE used to ignore missing values (NA) to prevent errors
summary_basketball #showing outputted frame
```

<div style="background-color: #f0f0f0; padding: 10px;">
    
#### Task 4.2: Determine the amount of missingness (2 points)

- Use your `dplyr` knowledge to create **one** table that calculates the number of missing values for each column in the data set (both numeric and categorical).
- Show the outputted table.

```r
missing_basketball <- tidy_basketball |> # making new table from tidy data
  summarise(across(everything(),  ~sum(is.na(.)))) # counting the number of missng values in every column

missing_basketball #showing outputted table
```

```r
# plotting a histogram showing the distribution of total win percentage across all teams
# theme, labs, fill, color, scale are used to enhance visualization
options(repr.plot.width = 8, repr.plot.height = 8) #setting plot size

histogram <- ggplot(tidy_basketball, aes(x=total_wins_pct))+ #setting plot and axis variable
    geom_histogram(bins=10, fill ="steelblue", color = "black")+ #plotting histogram, setting bin size, color fill and color outline
    labs(title = "Distribution of Team Winning Percentages", x = "Total Win Percentage (%)", y="Number of Teams")+ #setting axis and title labels
    theme(text = element_text(size = 14))+ #setting text size
    scale_x_continuous(breaks = seq(0, 100, by = 10)) + #setting x axis increments
    scale_y_continuous(breaks = seq(0, 600, by = 50)) #setting y axis increments
histogram #showing plot
```

```r
# plotting bar graph showing relative proportions of number of teams by tourney finish
# fill, color, scale, labs, theme all used to enhance visualization 

options(repr.plot.width = 9, repr.plot.height = 9) # setting plot size

bar_plot <- tidy_basketball |> #telling plot what data to use
  ggplot(aes(x = tourney_finish)) + #setting axis
  geom_bar(fill = "steelblue", color = "black") + #choosing bar chart, fill colors and outline colors
    scale_x_discrete(limits = c("opening_round_loss", "first_round_loss", "second_round_loss", "top_16_loss", "top_8_loss","top_4_loss", "top_2_loss", "champ"),labels = rev(c(
    "Champion","Runner-Up","Top Four","Top Eight", "Top Sixteen","Second Round Loss","First Round Loss", "Opening Round Loss")))+ #setting order of x axis variables and renaming variables
    labs(title = "Number of Teams by Tournament Finish", x = "Tournament Finish",y= "Number of Teams") + # setting title and axis titles
    theme(text = element_text(size = 15))+ #setting text size
    theme(axis.text.x = element_text(angle = 45, hjust = 1))+ #tilting x axis labels on a angle
    scale_y_continuous(breaks = seq(0, 1000, by = 100))# setting axis increments
bar_plot #showing plot
```

Visualization 1: 
This bar plot shows the average seed number for each tournament finish category. Lower seed numbers represent stronger teams (pre-tournament), so if teams that advance further have lower average seeds, seed may help predict tournament success. This is relevant because seed is one of the two predictor variables in the research question, while tournament finish is the response variable. This visualization shows that average seed number and tournament finish placement have a negative relationship (as seed number decreases, tournament placement advances). My next visualization will look at the second chosen predictor variable and how it correlates to the response variable.

```r
avg_seed <- tidy_basketball |> #create new table from tidy data
  group_by(tourney_finish) |> # group rows by tourney finish place
  summarise(mean_seed = mean(seed, na.rm = TRUE)) #calculate mean seed number for each group

visualization1 <- ggplot(avg_seed, aes(x = tourney_finish, y = mean_seed)) + #plot data, set x and y variables
  geom_col(fill = "steelblue", color = "black") + #set bar colors and outlines
  scale_x_discrete(limits = c("opening_round_loss", "first_round_loss", "second_round_loss", "top_16_loss", "top_8_loss","top_4_loss", "top_2_loss", "champ"),labels = rev(c(
    "Champion","Runner-Up","Top Four","Top Eight", "Top Sixteen","Second Round Loss","First Round Loss", "Opening Round Loss")))+ #setting order of variables and renaming variables
  labs(title = "Comparing Average Seeds of Tournament Placements",x = "Tournament Finish",y = "Average Seed Number") + #setting titles
  geom_text(aes(label = paste0(round(mean_seed, 1))),vjust = -0.4,size = 4) + # adding value labels to the bar graph
  theme(axis.text.x = element_text(angle = 45, hjust = 1))+ #tilting x axis labels
  theme(text = element_text(size = 15))+ #setting text size
  scale_y_continuous(breaks = seq(0, 12, by = 2))# setting y axis increments

visualization1
```
